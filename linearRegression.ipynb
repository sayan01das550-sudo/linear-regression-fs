## Class LinearRegression

```python
import matplotlib.pyplot as plt
class LinearRegression:
    X = [1, 2, 3, 4, 5, 6, 7, 8]
    Y = [2.9, 3.1, 4.0, 5.2, 6.1, 7.3, 8.0, 9.1]
    w,b = 0.0, 0.0
    lr = 0.01
    epochs = 100
```

## Method Gradient descent

```python
def GD(X, Y, w, b, lr, epochs):
    n = len(X)
    for epoch in range(epochs):
        dw, db = 0.0, 0.0
        for x,y in zip(X,Y):
            y_pred = w*x + b
            dw += (y_pred - y)*x
            db += (y_pred - y)
        w -= lr * dw/n
        b -= lr * db/n

    return w,b
```

## Main

```python
def main():
    model = LinearRegression()
    w,b = GD(model.X, model.Y, model.w, model.b, model.lr, model.epochs)
    print(f"Weight: {w}, Bias: {b}")
    
    line_x = [min(LinearRegression.X), max(LinearRegression.X)]
    line_y = [w * x + b for x in line_x]

    fig, ax = plt.subplots(figsize=(7, 4.5))

# Plot raw historical points
    ax.scatter(LinearRegression.X, LinearRegression.Y, color='darkorange', s=60, label='Actual Data', zorder=5)

# Plot fitted regression line
    ax.plot(line_x, line_y, color='royalblue', linewidth=2.5, label=f'Fit Line (y = {w:.2f}x + {b:.2f})')

# Personalize titles and layout
    ax.set_title("Linear Regression Fit via Gradient Descent", fontsize=12, fontweight='bold')
    ax.set_xlabel("X Features")
    ax.set_ylabel("Y Targets")
    ax.grid(True, linestyle='-.', alpha=0.5)
    ax.legend()

# 5. Display output
    plt.show()
```

```python
if __name__ == "__main__":
    main()
```
